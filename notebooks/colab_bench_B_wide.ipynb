```python
import os, subprocess, hashlib
from pathlib import Path
for v in ('OMP_NUM_THREADS','OPENBLAS_NUM_THREADS','MKL_NUM_THREADS',
          'NUMEXPR_NUM_THREADS','VECLIB_MAXIMUM_THREADS'):
    os.environ[v] = '1'
TAG = "phase3-freeze"
REPO = "https://github.com/hugogobato/scf-confounding-frontier.git"
if not Path("scf").exists():
    subprocess.run(["git", "clone", "--depth", "1", "-b", TAG, REPO, "scf"],
                   check=True)
os.chdir("scf")
def sha(p):
    return hashlib.sha256(Path(p).read_bytes()).hexdigest()[:16]
assert sha('code/de_formulas.py') == '5dffb441b6382d00', 'harness mismatch: code/de_formulas.py'
assert sha('code/simulator.py') == 'ef31ca2a201b7dc1', 'harness mismatch: code/simulator.py'
assert sha('code/detection.py') == '06586fe60b9f6573', 'harness mismatch: code/detection.py'
assert sha('code/estimators.py') == '7e27f25b2330b8e1', 'harness mismatch: code/estimators.py'
assert sha('code/benchmarks_data.py') == 'a4bd80f13fe03ca6', 'harness mismatch: code/benchmarks_data.py'
assert sha('code/fetch_bench_sources.py') == '9491e121c2f3338d', 'harness mismatch: code/fetch_bench_sources.py'
assert sha('code/benchmarks.py') == '5351edfdb2b689b0', 'harness mismatch: code/benchmarks.py'
assert sha('code/colab_shard_main.py') == '3a38d0734d9dbd3d', 'harness mismatch: code/colab_shard_main.py'
assert sha('code/bench_freeze.py') == '37cb235ef4b9c4a2', 'harness mismatch: code/bench_freeze.py'
assert sha('code/build_grids.py') == 'ca1bad75d189be16', 'harness mismatch: code/build_grids.py'
print("harness verified at tag", TAG)
```
